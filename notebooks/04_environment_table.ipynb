```python
import pandas as pd, matplotlib.pyplot as plt
from leptowatch.config import path
e = pd.read_csv(path('processed','env_weekly.csv'), parse_dates=['week_start'])
fig, ax = plt.subplots(3, 1, figsize=(12, 7), sharex=True)
for a in ['Kurunegala', 'Ratnapura']:
    g = e[e.rdhs == a]
    ax[0].plot(g.week_start, g.rain_mm, label=a); ax[1].plot(g.week_start, g.soil_m3m3)
    if 'paddy_flood' in g: ax[2].plot(g.week_start, g.paddy_flood)
ax[0].set_ylabel('rain mm/week'); ax[1].set_ylabel('soil m3/m3'); ax[2].set_ylabel('paddy flooded share'); ax[0].legend()
```
