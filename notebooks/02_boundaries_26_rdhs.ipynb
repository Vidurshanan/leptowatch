```python
import geopandas as gpd, matplotlib.pyplot as plt
from leptowatch.config import path
g = gpd.read_file(path('raw','boundaries','rdhs_26.geojson'))
ax = g.plot(column='provisional', edgecolor='white', figsize=(6,8), cmap='Set2', legend=True)
for _, r in g.iterrows(): ax.annotate(r.rdhs, r.geometry.representative_point().coords[0], fontsize=6, ha='center')
plt.title('26 RDHS areas (provisional = Ampara/Kalmunai split)')
```
